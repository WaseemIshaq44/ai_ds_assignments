## Write a function is_strong(password) that checks if the password:

* Has at least 8 characters

* Contains numbers

* Contains uppercase letters

* Contains lowercase letters

### Return True or False.

```python
def is_strong(password):
    if len(password)>=8:
        up = False
        lo = False
        dig = False
        for char in password:
            if char.isupper():
                up = True
        for char in password:
            if char.islower():
                lo = True
        for char in password:
            if char.isdigit():
                dig = True
        if up and lo and dig:
            return True
        else:
            return False
    else:
        return False
```

## Write a function extract_username(email) that returns everything before "@".

### Example: "ubaid@gmail.com" → "ubaid"

```python
def extract_username(email):
    user_name=''
    for char in email:
        if char == '@':
            break
        else:
            user_name+=char
    return user_name
```

## Write a function process_transaction(balance, amount, transaction_type) that supports:

* "deposit" → add amount

* "withdraw" → subtract amount

* "transfer" → subtract amount + 1% transfer fee

### Return the updated balance.
### If amount > balance (for withdraw/transfer), return "Transaction Failed".

```python
def process_transaction(balance, amount, transaction_type):
    message = ''
    if amount>balance:
        message += 'Transaction Failed'
    else:
        if transaction_type =='deposit':
            balance+=amount
            message +=  f'''
            _________________________________________________________________________________________
        
            balance: {balance}                                               deposited amount:{amount}
            __________________________________________________________________________________________'''
        elif transaction_type =='withdraw':
            balance-=amount
            message+= f'''
            _________________________________________________________________________________________
            
            balance: {balance}                                               withdraw amount:{amount}
            __________________________________________________________________________________________'''
        elif transaction_type == 'transfer':
            balance -=(amount + amount*.01)
            message+= f'''
            _________________________________________________________________________________________
            
            balance: {balance}                                               transfered amount:{amount}
    
            transfer fee: {amount*.01}
            __________________________________________________________________________________________'''
            
    return print(message)
```

```python
process_transaction(2000, 500, 'transfer')
```

```python
process_transaction(500, 2000, 'transfer')
```

## Airline Seat Booking With Constraints

### Write a function book_seat(seats, seat_requested):

### Constraints:

    If seat_requested not in seats → return "Invalid Seat"
    
    If seat is "booked" → return "Seat Already Booked"
    
    Otherwise book the seat and return updated seats dictionary

### Example seats dictionary:
{
  "1A": "free",
  "1B": "booked",
  "2A": "free"
}

```python
def book_seat(seats, seat_requested):
    message = ''
    if seat_requested in seats and seats[seat_requested]=='booked':
        message += f'Seat is Already Booked'
    elif seat_requested not in seats:
        message+= f'Invalid Seat'
    else:
        seats[seat_requested]=='booked'
        message+= f'seat booked successfully'
    return message
```

```python
book_seat({ "1A": "free", "1B": "booked", "2A": "free" }, '2A')
```

        "name" → employee name
        
        "basic" → basic salary
        
        "overtime_hours" → extra hours worked
        
        "performance" → string ("excellent", "good", "average", "poor")

### The function must:

#####    ✔ 1. Calculate Gross Salary
    
    Gross = basic salary
    
    overtime pay
    
    performance bonus
    
#####    ✔ 2. Overtime Pay Rules
    
    1 overtime hour = 300 PKR
    
####    ✔ 3. Performance Bonus Rules
    
    excellent → 20% of basic
    
    good → 10%
    
    average → 5%
    
    poor → 0%
    
#####    ✔ 4. Apply Deductions
    
    Income tax = 12% of gross
    
    Medical insurance = 1000 PKR
    
#####    ✔ 5. Return & print a full salary slip with this format:
        Employee: Ali
        Basic Salary: 50000
        Overtime Pay: 6000
        Performance Bonus: 10000
        Gross Salary: 66000
        Income Tax (12%): 7920
        Medical Insurance: 1000
        Net Salary: 57080

### 📝 Example Input:
    employee = {
        "name": "Ali",
        "basic": 50000,
        "overtime_hours": 20,
        "performance": "excellent"

    Employee: Ali
    Basic Salary: 50000
    Overtime Pay: 6000
    Performance Bonus: 10000
    Gross Salary: 66000
    Income Tax (12%): 7920
    Medical Insurance: 1000
    Net Salary: 57080

## An online store calculates final prices dynamically:

### Function calculate_discount(price, category) → category-based discounts:

    Electronics > $1000 → 15%
    
    Clothing > $500 → 10%
    
    Groceries → 5%
    
    Function apply_coupon(price, coupon_code) → nested function applies coupon:
    
    SAVE10 → $10 off
    
    SAVE20 → $20 off if price > $200
    
    Function final_price(price, category, coupon) → combines discount + coupon + tax (8%)

### 📥 Input Example:
    final_price(1200, "Electronics", "SAVE10")

### 📝 Example Output:
    Price after discount: 1020
    Price after coupon: 1010
    Final Price with Tax: 1090.8

```python
def calculate_discount(price, category, coupon_code):
    discount = 0  
    if category == 'Electronics' and price >1000:
        discount += price*.15
    elif category == 'Clothing' and price >500:
        discount += price*.10
    elif category == 'Groceries':
        discount += price*.05
    def apply_coupon(price, coupon_code):
        coupon = 0
        if coupon_code =='SAVE10':
            coupon += 10
        elif coupon_code =='SAVE20' and price>200:
            coupon += 20
        return coupon
    price_after_dicount = price - discount
    price_after_coupon = price_after_dicount - apply_coupon(price,coupon_code)
    final_price_with_tax = price_after_coupon+ price_after_coupon*.08
    return print(f'''
    Original Price : {price}
    Price after discount: {price_after_dicount}
    Price after coupon: {price_after_coupon}
    Price Price with tax: {final_price_with_tax}''')        
```

```python
calculate_discount(1200,'Groceries','SAVE20')
```

```python
def room_charges(nights, room_type):
    charge = 0
    if room_type == 'Standard':
        charge += 100*nights
    elif room_type == 'Deluxe':
        charge += 200*nights
    elif room_type == 'Suit':
        charge += 400*nights
    return charge
def additional_services(services):
    add_serv = 0
    if 'Breakfast' in services:
        add_serv+=20*services.count('Breakfast')
    if 'Spa' in services:
        add_serv +=50*services.count('Spa')
    if 'Gym' in services:
        add_serv +=10*services.count('Gym')
    return add_serv
def apply_membership_discount(amount, membership_years):
    membership_discount=0
    if membership_years > 5:
        membership_discount += amount*.10
    elif membership_years>2:
        membership_discount += amount*.05
    return membership_discount
def total_bill(nights, room_type, services, membership_years):
    room_chrgs = room_charges(nights,room_type)
    add_srvs = additional_services(services)
    amount = room_charges(nights,room_type)+additional_services(services)
    mmbrship_dcont = apply_membership_discount(amount,membership_years)
    return print(f'''
    Room Charges: {room_chrgs}
    Additional Services: {add_srvs} 
    Membership Discount: {mmbrship_dcont} 
    Total Amount: {room_chrgs+add_srvs-mmbrship_dcont}
    ''')
        
```

## A school calculates student performance:

### Each student has marks in Math, Science, English, History

    Function average_score(marks) → calculates average
    
    Function grade(avg) → nested function assigns grade:
    
    ≥90 → A, 80–89 → B, 70–79 → C, 60–69 → D, <60 → F
    
    Function honor_roll(avg, extracurricular) → gives "Honor Roll" if avg ≥85 and extracurricular ≥3
    
    Function student_summary(student) → prints name, average, grade, honor roll status

### 📥 Input Example:
    student = {"name":"Ali", "marks":[85,90,80,95], "extracurricular":3}
    student_summary(student)

### 📝 Example Output:
    Name: Ali
    Average Score: 87.5
    Grade: B
    Honor Roll: Yes

```python
def average_score(marks, extracurricular):
    ext =extracurricular
    average=0
    average += sum(marks)/len(marks)
    def grade(avg):
        if avg>=90:
            return 'A'
        elif avg>=80:
            return 'B'
        elif avg>=70:
            return 'C'
        elif avg>=60:
            return 'D'
        else:
            return 'F'
    grd = grade(average)
    def honor_roll(avg, extracurricular):
        honr_rool =''
        if avg>=85 and extracurricular>=3:
            honr_rool += 'Honor Roll'
            return honr_rool
    hnr = honor_roll(average,ext)
    return average,  grd,  hnr 
def student_summary(stud_dict):
    sumary = average_score(stud_dict['marks'],stud_dict['extracurricular'])
    return print(f'''
    Name: {stud_dict['name']}
    Average Score: {sumary[0]}
    Grade: {sumary[1]}
    Honor Roll: {sumary[2]}''')
```
