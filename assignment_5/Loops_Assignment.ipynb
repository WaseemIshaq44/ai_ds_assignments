## Create a menu-driven banking system using loops:
* Create Account
* Deposit
* Withdraw
* Check Balance
* Exit

```python
print(f'''
______________________________________________________________________________________
______________________________________________________________________________________
             Enter your Choice
             1. Create Account
             2. Deposit
             3. Withdraw
             4. Check Balance
             5. Exit
______________________________________________________________________________________
______________________________________________________________________________________
''')
accounts = []
import random
for i in range(20):
    user_choice = int(input('Enter your choice: '))
    if user_choice==1:
        name = input('Enter the name: ')
        age = input('Enter the age: ')
        balance = int(input('Enter the initial deposit'))
        account_number = random.randint(4444,9999)
        user_account = {'account_number':account_number,'name':name, 'age':age, 'balance':balance}
        accounts.append(user_account)
        print(f'Mr {name} your Account is created successfully. Your account number is {account_number} and your balance is {balance}')
    elif user_choice ==2:
        user_account_number = int(input('Enter your Account Number: '))
        for i in accounts:
            if i['account_number'] ==user_account_number:
                deposit_amount = int(input('Enter the deposit amount: '))
                if deposit_amount>0:
                    i['balance']+=deposit_amount
                    print(f'{i['name']} the amount of {deposit_amount} is successfully deposited in your account number {i['account_number']}. Now your new balance is {i['balance']}')
                    break
                else:
                    print(f'{i['name']} the amount should be positive')
                    break
        else:
            print('Account not found')
    elif user_choice ==3:
        user_account_number = int(input('Enter your Account Number: '))
        for i in accounts:
            if i['account_number'] ==user_account_number:
                withdraw_amount = int(input('Enter the amount you want to withdraw: '))
                if withdraw_amount>0:
                    if withdraw_amount>i['balance']:
                        print(f'Insufficient Balance: You cannot withdraw {withdraw_amount} from {i['balance']} ')
                        break
                    else:
                        i['balance']-=withdraw_amount
                        print(f'{i['name']} the amount of {withdraw_amount} is successfully withdraw from your account number {i['account_number']}. Now your new balance is {i['balance']}')
                        break
                else:
                    print(f'{i['name']} the amount should be positive')
                    break
        else:
            print('Account not found')
    elif user_choice ==4:
        user_account_number = int(input('Enter your Account Number: '))
        for i in accounts:
            if i['account_number'] ==user_account_number:
                print(f'''
__________________________________________________________________________
__________________________________________________________________________
                     Account Detail
_________________________________________________________________________
__________________________________________________________________________
        Name :                         {i['name']}
        Account Number:                {i['account_number']}
        Age:                           {i['age']}
        Balance:                       {i['balance']}
__________________________________________________________________________
__________________________________________________________________________
''')
                break
        else:
            print('Account not found')
    elif user_choice ==5:
        print('Thanks for using our App')
        break             
```

## Shop Sales Tracker
### Goal:

    Track daily sales of 3 products over 5 days and calculate the total revenue.

### Steps:

    Outer loop → iterate over 5 days.
    
    Inner loop → take sales amount for 3 products each day.
    
    Store daily total in a list.
    
    If any product sale = 0 → continue (skip that product).
    
    Print all daily totals and grand total.

### Example:
    Day 1:
    Product 1: 200
    Product 2: 150
    Product 3: 0 → skipped
    Day 2:
    Product 1: 300
    Product 2: 250
    Product 3: 400
    → Daily Totals = [350, 950], Grand Total = 1300

```python
daily_totals = []
for i in range(5):
    print(f'Day {i+1}: ')
    price=0
    for j in range(3):
        user_price =int(input(f'Product {j+1}: '))
        if user_price==0:
            continue
        else:
            price+=user_price
    daily_totals.append(price)
grand_total = sum(daily_totals)
print(f'Daily Totals = {daily_totals} Grand Total = {grand_total}')        
```

## Invoice Tax Calculator

Take multiple items in multiple invoices and calculate tax per invoice.

## Student Result Sheet
### Goal:

    Generate report cards for each student in a class.

### Steps:

    Outer loop → 5 students.
    
    Inner loop → 4 subjects.

    If marks < 33 → print “Failed in Subject”.

    Calculate total and grade per student.

### Example:
    Student 1: [80, 90, 70, 85] → Grade: A  
    Student 2: [50, 32, 45, 40] → Failed in Subject 2

```python
report_card = []
for i in range(5):
    student_name = input('Enter the student Name: ')
    student_marks =[]
    obtain_marks = 0
    percentage = 0
    remarks =''
    failed_subject = 0
    for j in range(4):
        marks = int(input(f'Enter the marks of Subject {j+1}'))
        if marks<33:
                failed_subject+=1
                remarks += f'Failed in subject {j+1}.   '
        student_marks.append(marks)
    obtain_marks = sum(student_marks)
    percentage = (obtain_marks/(len(student_marks)*100))*100
    
    if percentage >=80:
        grade='A'
    elif percentage>=70:
        grade = 'B'
    elif percentage>=60:
        grade = 'C'
    elif percentage>=50:
        grade = 'D'
    elif percentage >= 33:
        grade = 'Pass'
    else:
        grade = 'Fail'    

    if failed_subject>0:
        report_card.append({'Student Name':student_name,'Student Marks':student_marks, 'Obtain Marks':obtain_marks, 'Total Marks':len(student_marks)*100, 'Percentage': percentage, 'Grade': remarks })
    else:
        remarks+='Pass in all subjects'
        report_card.append({'Student Name':student_name,'Student Marks':student_marks, 'Obtain Marks':obtain_marks, 'Total Marks':len(student_marks)*100, 'Percentage': percentage, 'Grade':grade })

for i in report_card:
    print(f'''
_________________________________________________________________________________________________

{i['Student Name']}    :    {i['Student Marks']}
Grade    :    {i['Grade']}

__________________________________________________________________________________________________
''')
```

## Holiday Package Evaluator
### Goal:

Compare multiple travel packages across features like hotel, transport, food.

### Steps:

Outer loop → 3 packages.

Inner loop → 3 features.

Input rating out of 5.

Skip if rating = 0.

Print average rating per package.

### Example:
Package 1: [4, 5, 5] → Avg = 4.67  
Package 2: [3, 4, 0] → Avg = 3.5

```python
hotel_pakage =[]
statement = ''
for i in range(3):
    print(f'Package {i+1}:')
    review = []
    for j in range(3):
        if j==0:
            hotel_review = int(input('Enter the rating for hotel (1-5): '))
            review.append(hotel_review)
        elif j==1:
            transport_review = int(input('Enter the rating for transportation (1-5): '))
            review.append(transport_review)
        elif j==2:
            food_review = int(input('Enter the rating for foor (1-5): '))
            review.append(food_review)
    sum_valid_review = 0
    num_valid_review = 0
    for k in review:
        if k == 0:
            continue
        else:
            sum_valid_review+=k
            num_valid_review +=1
    average = sum_valid_review/num_valid_review
    hotel_pakage.append({f'Package {i+1}': review, 'Average': average})
    statement +=f'Package {i+1}: {review}   Average= {average}\n'
print(statement)
```
