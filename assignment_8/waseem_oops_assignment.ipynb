2. Write a Python program to create a class called 'Rectangle' with properties for
width and height. Include two methods to calculate rectangle area and perimeter.
Create an instance of the 'Rectangle' class and calculate its area and perimeter.

```python
class rectangle():
    def __init__(self, width, height):
        self.width = width
        self.height = height
    def area_of_rectangle(self):
        area = self.width*self.height
        return f'Area of the rectange with length {self.width} and height {self.height} is {area}'
    def perimeter_of_rectange(self):
        perimeter = 2*(self.width+self.height)
        return f'perimeter of the rectange with length {self.width} and height {self.height} is {perimeter}'
```

```python
rect1 = rectangle(24,26)
```

```python
rect1.area_of_rectangle()
```

```python
rect1.perimeter_of_rectange()
```

4. Write a Python program that creates a class called "BankAccount" with 
properties for account number and balance. Include methods to deposit and 
withdraw money from the account. Create some instances of the "BankAccount" 
class, deposit some money, and withdraw a portion of it. 

```python
class BankAccount():
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self._balance = balance
    def deposit(self, amount):
        if amount<0:
            return 'Your cannot deposite negative amount'
        else:
            self._balance+=amount
            return f'{amount} is deposited successfully'
    def withdraw(self, amount):
        if amount>self._balance:
            return 'Insufficent balance'
        if amount<0:
            return 'Your cannot withdraw negative amount'
        else:
            self._balance-=amount
            return f'{amount} is withdraw successfully'
```

5. Write a Python program that creates a class called 'Shape' with a method to 
calculate the area. Create two subclasses, 'Circle' and 'Triangle', that inherit from 
the 'Shape' class and override the area calculation method. Create an instance of 
the 'Circle' class and calculate its area. Similarly, do the same for the 'Triangle' 
class. 

```python
class Shape():
    def __init__(self, length, width,):
        self.length = length
        self.width = width
    def area(self):
        self.area = self.length *self.width
```

```python
class circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        area = 3.14*self.radius**2
        return f'The area of circle with radius {self.radius} is {area}'
```

```python
class Triangle(Shape):
    def __init__(self, base,height):
        self.base = base
        self.height = height
    def area(self):
        self.area = (1/2)*(self.base*self.height)
        return f'The area of triangle with base {self.base} and height {self.height} is {self.area}'
```

```python
cir = circle(55)
```

```python
cir.area()
```

```python
tri = Triangle(5,3)
```

```python
tri.area()
```

10. Write a Python program that creates a class called Product with properties for 
product ID, name, and price. Include a method to calculate the total price by 
multiplying the price by the quantity. Create a subclass called 
PersonalCareProduct that inherits from the Product class and adds an additional 
property for the warranty period. Override the total price calculation method to 
include the warranty period. Create an instance of the PersonalCareProduct 
class and calculate its total price. 

```python
class Product():
    def __init__(self,product_id,name,price):
        self.product_id = product_id
        self.name = name
        self.price = price
    def calculate_total_price(self,quantity):
        self.total_price = self.price *quantity
        return f'The Total price of {quantity} numbers of {self.name} is {self.total_price}'
```

```python
class PersonalCareProduct(Product):
    def __init__(self,product_id,name,price, warranty_period):
        super().__init__(product_id, name, price)
        self.warranty_period = warranty_period
    def calculate_total_price(self, quantity):
        self.total_price = self.price*quantity +self.warranty_period*10
        return f'The total price of {quantity} {self.name} with {self.warranty_period} months of warranty is {self.total_price}'    
```

```python
prod1 = PersonalCareProduct(181,'Almera',18000,12)
```

```python
prod1.calculate_total_price(4)
```

```python
prodec = Product(190,'Shoes',1500)
```

```python
prodec.calculate_total_price(3)
```

11. Write a Python program that creates a class called BankAccount with 
properties for account number, account holder name, and balance. Include 
methods to deposit, withdraw, and transfer money between accounts. Create 
multiple instances of the BankAccount class and perform operations such as 
depositing, withdrawing, and transferring money. 

```python
import random
```

```python
class BankAccount():
    def __init__(self,name, initial_deposit):
        self.account_number= random.randint(4444,9999)
        self.name = name
        self.balance = initial_deposit
    def deposit(self, account_number, amount):
        if amount>0:
            self.balance+=amount
            return f'Mr. {self.name} Rs: {amount} is deposited in your account number {self.account_number}. Now your total balance is {self.balance}'
        else:
            return f'You Entered {amount}. You cannot deposite the negative amount'
    def withdraw(self, account_number,amount):
        if amount>=self.balance:
            self.balance-=amount
            return f'Mr. {self.name} Rs: {amount} is withdraw from your account number {self.account_number}. Now your total balance is {self.balance}'
        else:
            return f'You cannot withdraw this amount {amount}'
    def transfer(self,other,amount):
        self.balance-=amount
        other.balance+=amount
        return f'''Rs {amount} transferred from 
                {self.name} ({self.account_number}) to 
                {other.name} ({other.account_number})'''
```
