## Two team player sets given. Find common players and unique players

```python
team1 = { "Noman", "Naqash", "Atiq", "Areeb"}
team2 = {"Saleem", "Naqash","Wajahat", "Noman" }
print(f"The unique players in both teams are {team1.symmetric_difference(team2)}")
print(f"The common players in both teams are {team1.intersection(team2)}")
```
